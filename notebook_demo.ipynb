# 🎨 Lightshow Interactive Notebook

This notebook demonstrates the Jupyter integration for the `lightshow` Art-Net lighting library.

* **Single Import**: `from lightshow.jupyter import *` re-exports all primitives + adds rich visual inspection.
* **Authentic 2D Stage Visualizer**: Renders the exact physical fixture faces (e.g. EUROLITE PARty Spot with 2 concentric rings of 3 and 9 LEDs).
* **Interactive Time Scrubber**: Scrub animations across 2D space right in your browser.
* **Non-Blocking Live Streaming**: `stream(scene)` streams at 40Hz in a background thread with live hot-swapping.


```python
# 1. Single convenience import: re-exports all lightshow tools + Jupyter visualizers
from lightshow.jupyter import *
from my_rig import PARS, STROBE_BAR, STAGE_RGB, STAGE, create_rig

print("✓ Lightshow Jupyter & Stage modules loaded successfully!")
```

## 1. Visualizing Clips with Inline SVG Ribbons
Simply evaluate any `Clip` in a cell to render its interactive timeline strip across fixtures:

```python
# A rainbow wave staggered across 4 fixtures over 4.0 seconds
wave = Rainbow(duration=4.0).spread()
wave
```

```python
# A pulsating amber flame with occasional white strobe flashes
flame = Pulse(Color(1.0, 0.3, 0.0), duration=0.8) >> Hold(Color.WHITE, w=1.0, strobe=0.9, duration=0.2)
flame * 3
```

## 2. Multi-Fixture Scene Composition & Timeline Tracks
When you compose fixtures together into a `Scene`, evaluating it renders the DMX output tracks for every fixture group:

```python
# Combine PARs and the 4-segment Strobe Bar
halloween_scene = (
    PARS(Gradient([Color(0.8, 0.0, 0.0), Color(1.0, 0.25, 0.0)], duration=3.0).spread())
    | STROBE_BAR(Chase(Color.CYAN, step_duration=0.15, tail=1.5).pingpong() @ 3.0)
)
halloween_scene
```

## 3. 2D Virtual Stage (Authentic Physical Fixture Faces!)
`STAGE` defines the physical geometry of your living room rig:
* **4x EUROLITE LED PARty Spot**: Rendered with their authentic 2 concentric rings (3 inner LEDs + 9 outer LEDs), optical white desaturation, and strobe bloom.
* **1x EUROLITE LED BCL-40 Strobe Bar**: Rendered with its 4 segment compartments and bi-color SMD white chips.

```python
# Render a snapshot of the stage at t = 1.5s
STAGE(halloween_scene, t=1.5)
```

```python
# Interactive 2D Stage Scrubber: drag slider or hit Play to preview live!
STAGE.inspect(halloween_scene, duration=3.0, fps=15.0)
```

## 4. Non-Blocking Live Streaming & Hot-Reloading
Run `stream(scene)` to stream to your Art-Net gateway (`192.168.1.158`) in a background thread.
The notebook kernel remains unblocked—you can change code in another cell, call `stream(new_scene)`, and it hot-swaps live without dropping frames!

```python
# Start background streaming (40 Hz DMX to 192.168.1.158)
# Set dry_run=True if you just want to test without broadcasting UDP
rig = create_rig(dry_run=True)  # change to dry_run=False for live lights!

s = stream(halloween_scene * inf, rig=rig)
s
```

```python
# Hot-swap to a new look live on the fly!
synthwave = (
    PARS(Pulse(Color.MAGENTA, duration=1.0).spread())
    | STROBE_BAR(Hold(Color.CYAN, duration=1.0))
)
stream(synthwave * inf)
```
